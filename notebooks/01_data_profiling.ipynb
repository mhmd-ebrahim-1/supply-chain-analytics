# 01. Data Discovery & Profiling
## DataCo Smart Supply Chain Dataset
This notebook performs initial schema inspection, data type detection, key analysis, missing value auditing, and data quality validation.

```python
import pandas as pd
import numpy as np
import os

raw_path = '../data/raw/DataCoSupplyChainDataset.csv'
df = pd.read_csv(raw_path, encoding='latin1')
print(f'Loaded {len(df):,} rows and {len(df.columns)} columns.')
```

```python
print('--- Dataset Information ---')
df.info()
```

```python
print('--- Grain & Entity Key Uniqueness ---')
print('Order Item Id unique:', df['Order Item Id'].nunique())
print('Order Id unique:', df['Order Id'].nunique())
print('Product Card Id unique:', df['Product Card Id'].nunique())
print('Customer Id unique:', df['Customer Id'].nunique())
```

```python
print('--- Missing Values Analysis ---')
nulls = df.isnull().sum()
null_report = pd.DataFrame({'null_count': nulls, 'null_pct': (nulls / len(df) * 100).round(2)})
null_report[null_report['null_count'] > 0]
```

```python
print('--- Delivery Status Distribution ---')
print(df['Delivery Status'].value_counts(dropna=False))
```

```python
print('--- Date Coverage ---')
print('Order Date:', df['order date (DateOrders)'].min(), 'to', df['order date (DateOrders)'].max())
print('Shipping Date:', df['shipping date (DateOrders)'].min(), 'to', df['shipping date (DateOrders)'].max())
```
